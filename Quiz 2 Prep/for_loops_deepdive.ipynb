## 1. The 5 Types of For Loops — Overview

There are **5 main forms** of for loops in this course. Here's a quick summary — the rest of this notebook explains each one in depth.

| # | Syntax | Use when... |
|---|--------|-------------|
| 1 | `for val in my_list:` | You just need the VALUES, no modification |
| 2 | `for i in range(len(my_list)):` | You need to MODIFY the list, or use the index |
| 3 | `for i in range(n):` | You want to repeat something n times |
| 4 | `for a, b in zip(list1, list2):` | You need to loop over TWO+ lists simultaneously |
| 5 | Nested loops | You need a loop inside a loop |

## 2. Type 1: Loop by Value
### `for val in my_list:`

**Use this when:** You only need to READ or USE the values — you do NOT need to change the list.

The loop variable (`val`) automatically takes the VALUE of each element, one at a time.

````python
car_list = ["KIA", "Ferrari", "Ford", "Tesla"]

# The variable 'car_name' takes each value in the list, one by one
for car_name in car_list:
    print("Dear customer, we are writing about your " + car_name + " vehicle.")
````

````python
# What this loop is doing, step by step ("unrolled"):
car_list = ["KIA", "Ferrari", "Ford", "Tesla"]

car_name = car_list[0]   # "KIA"
print("Dear customer, we are writing about your " + car_name + " vehicle.")

car_name = car_list[1]   # "Ferrari"
print("Dear customer, we are writing about your " + car_name + " vehicle.")

car_name = car_list[2]   # "Ford"
print("Dear customer, we are writing about your " + car_name + " vehicle.")

car_name = car_list[3]   # "Tesla"
print("Dear customer, we are writing about your " + car_name + " vehicle.")
````

````python
# ✅ GOOD USES of 'loop by value':

# 1. Print every item
movies = ["Rear Window", "Inception", "The Wailing"]
for movie in movies:
    print(movie)

print()

# 2. Sum all values
ratings = [7.8, 6.3, 8.5, 7.4]
total = 0
for r in ratings:
    total = total + r
print("Sum of ratings:", total)

print()

# 3. Apply a condition to each value
for r in ratings:
    if r >= 8.0:
        print(r, "is a great rating!")
````

````python
# ❌ LIMITATION: You CANNOT modify the original list this way

numbers = [1, 2, 3, 4, 5]

for val in numbers:
    val = val * 10      # This only changes the LOCAL variable 'val'
                        # The original list is NOT changed!

print(numbers)          # Still [1, 2, 3, 4, 5] — unchanged!
````

**Key insight:** `val` is just a temporary copy of each element. Changing `val` does NOT change the list.

If you need to modify the list → use **Type 2** (loop by index).

## 3. Type 2: Loop by Index
### `for i in range(len(my_list)):`

**Use this when:**
- You need to **modify** elements of the list
- You need to know the **position** (index) of each element
- You need to access elements from **two lists at the same time** using the same index

`range(len(my_list))` produces: 0, 1, 2, ..., len-1 — exactly all the valid indices.

````python
# Understand what range(len(...)) gives you:
my_list = ["a", "b", "c", "d"]

print("len(my_list) =", len(my_list))                  # 4
print("range(len(my_list)) =", list(range(len(my_list))))  # [0, 1, 2, 3]
print("These are exactly the valid indices of my_list!")
````

````python
# ✅ Modifying a list in-place (requires index loop)
numbers = [1, 2, 3, 4, 5]

for i in range(len(numbers)):
    numbers[i] = numbers[i] * 10   # This DOES modify the original list

print(numbers)   # [10, 20, 30, 40, 50] — modified!
````

````python
# ✅ Replace out-of-range values with None (from the practice quiz!)
temp_data = [-5, -2, 0, 1000, 2, -12, 44, 40, 30, -10]
min_val = -10
max_val = 40

for i in range(len(temp_data)):
    if temp_data[i] > max_val or temp_data[i] < min_val:
        temp_data[i] = None   # Replace the actual element

print(temp_data)
````

````python
# ✅ Using the index to print position information
car_list = ["KIA", "Ferrari", "Ford", "Tesla"]

for i in range(len(car_list)):
    print("Position", i+1, "on the waitlist:", car_list[i])
````

````python
# ✅ Access TWO lists by the same index
titles  = ["Umbrellas of Cherbourg", "M3gan", "Psycho"]
ratings = [7.8, 6.3, 8.5]

for i in range(len(titles)):
    title  = titles[i]    # use i to grab from BOTH lists
    rating = ratings[i]
    print(title, "->", rating)
````

**Note:** When you just need values from two lists (without modifying them), `zip()` is cleaner. But when you **need the index** for other reasons (printing position, modifying one of the lists, etc.), use `range(len(...))` .

## 4. Type 3: Loop with `range()` Only
### `for i in range(n):`

**Use this when:** You want to **repeat something n times**, often to run a simulation or fill a pre-allocated array.

You're not necessarily looping over a list — you're just counting iterations.

````python
# range() variants — study these!

print("range(5):",       list(range(5)))        # 0, 1, 2, 3, 4
print("range(1, 6):",    list(range(1, 6)))      # 1, 2, 3, 4, 5
print("range(1, n+1):",  list(range(1, 8)))      # 1, 2, 3, ..., 7  (when n=7)
print("range(2, 10, 2):",list(range(2, 10, 2)))  # 2, 4, 6, 8  (step by 2)
````

````python
# ✅ Sum of first n integers
n = 10
total = 0

for i in range(1, n + 1):   # i goes 1, 2, 3, ..., 10
    total = total + i

print("Sum:", total)   # Should be 55
print("Check: n*(n+1)/2 =", n*(n+1)//2)  # Formula
````

````python
# ✅ Run a simulation 2000 times (the CLT simulation from Lecture 9)
num_simulations = 2000
sample_size = 50

vec_xbar = np.zeros(num_simulations)   # pre-allocate results array

for iteration in range(num_simulations):   # repeat 2000 times
    sample = np.random.uniform(low=-2, high=2, size=sample_size)
    vec_xbar[iteration] = sample.mean()   # store result at index 'iteration'

plt.hist(vec_xbar)
plt.title("Distribution of sample means (n=" + str(sample_size) + ")")
plt.xlabel("Sample mean")
plt.ylabel("Frequency")
plt.show()
````

## 5. Type 4: Loop with `zip()`
### `for a, b in zip(list1, list2):`

**Use this when:** You want to loop over **two or more lists at the same time**, and you don't need the index.

`zip()` pairs up elements by position. On each iteration you get one element from EACH list.

````python
# The three approaches to looping over two lists — compared

titles  = ["Umbrellas of Cherbourg", "M3gan", "Psycho"]
ratings = [7.8, 6.3, 8.5]

# Approach A: manual index variable (works but clunky)
print("--- Approach A ---")
index = 0
for title in titles:
    rating = ratings[index]
    print(title, "->", rating)
    index = index + 1

print()

# Approach B: range(len(...))  (works, but more typing)
print("--- Approach B ---")
for i in range(len(titles)):
    print(titles[i], "->", ratings[i])

print()

# Approach C: zip  (cleanest — preferred when no modification needed)
print("--- Approach C: zip (BEST) ---")
for title, rating in zip(titles, ratings):
    print(title, "->", rating)
````

````python
# zip over 3 or 4 lists — just add more!
titles    = ["Umbrellas of Cherbourg", "M3gan", "Psycho"]
ratings   = [7.8, 6.3, 8.5]
years     = ["1964", "2022", "1960"]
directors = ["Jacques Demy", "Gerard Johnstone", "Alfred Hitchcock"]

for title, rating, year, director in zip(titles, ratings, years, directors):
    msg = "'" + title + "' (" + year + ") - dir. " + director + " | IMDB: " + str(rating)
    print(msg)
````

````python
# zip works with numpy arrays, tuples, and lists mixed together
title_list   = ["Umbrellas of Cherbourg", "M3gan", "Psycho"]
rating_array = np.array([7.8, 6.3, 8.5])   # numpy array
year_tuple   = ("1964", "2022", "1960")     # tuple

for title, rating, year in zip(title_list, rating_array, year_tuple):
    print(title, year, rating)
````

````python
# ⚠️ IMPORTANT: zip stops at the SHORTEST list
list_long  = [1, 2, 3, 4, 5]
list_short = ["a", "b", "c"]

for num, letter in zip(list_long, list_short):
    print(num, letter)

# Output: only 3 iterations (stops when shortest list is exhausted)
````

````python
# ✅ zip for the subplot trick (Lecture 9)
# Instead of using indices to access 3 lists, zip them!
sample_sizes = [1, 10, 50, 100, 150, 200]
fig_rows     = [0, 0, 0, 1, 1, 1]
fig_cols     = [0, 1, 2, 0, 1, 2]

fig, subfig_array = plt.subplots(2, 3, figsize=(9, 6))

for sample_sz, row, col in zip(sample_sizes, fig_rows, fig_cols):
    # one iteration gives us: sample_sz, its row, AND its column — all at once!
    vec_xbar = np.zeros(500)
    for i in range(500):
        vec_xbar[i] = np.random.uniform(-2, 2, size=sample_sz).mean()
    
    subfig_array[row, col].hist(vec_xbar)
    subfig_array[row, col].set_title("n = " + str(sample_sz))

plt.tight_layout()
plt.show()
````

## 6. Type 5: Nested Loops
### A loop inside another loop

**Use this when:** For each iteration of the outer loop, you need to do a whole series of repeated steps.

**How it works:**
- The **outer loop** runs once per "big step" (e.g. each sample size)
- The **inner loop** runs completely (all its iterations) for EACH step of the outer loop

````python
# Simple example to understand nesting — trace this by hand!
for outer in ["A", "B", "C"]:
    for inner in [1, 2, 3]:
        print(outer, inner)
    print("--- (inner loop finished for outer =", outer, ")")
````

````python
# Real use case: run 2000 simulations for EACH of several sample sizes
sample_size_list = [1, 10, 100]
num_simulations = 1000

for sample_sz in sample_size_list:               # outer: each sample size
    vec_xbar = np.zeros(num_simulations)
    
    for iteration in range(num_simulations):      # inner: 1000 simulations
        sample = np.random.normal(size=sample_sz)
        vec_xbar[iteration] = sample.mean()
    
    # This runs once per sample size, AFTER the inner loop completes
    print(f"n={sample_sz}: mean of xbar = {vec_xbar.mean():.4f}, std = {vec_xbar.std():.4f}")
````

## 7. Loop Pattern: Accumulators

An **accumulator** is a variable that collects a running result across loop iterations.

**Three golden rules:**
1. Initialize the accumulator **BEFORE** the loop
2. Update it **INSIDE** the loop
3. Use the final result **AFTER** the loop

````python
# ✅ CORRECT accumulator pattern
data = [5, 3, 8, 2, 10, 7]

total = 0        # ① Initialize BEFORE loop
count = 0

for val in data:
    total += val  # ② Update INSIDE loop
    count += 1

avg = total / count  # ③ Use result AFTER loop
print("Average:", avg)
````

````python
# ❌ WRONG: accumulator initialized inside the loop
data = [5, 3, 8, 2, 10, 7]

for val in data:
    total = 0    # WRONG! Resets to 0 at the start of EVERY iteration
    total += val

print("Wrong total:", total)   # Only has the LAST value (7), not the sum!
````

````python
# ❌ WRONG: computing average inside the loop
data = [5, 3, 8, 2, 10, 7]
total = 0
count = 0

for val in data:
    total += val
    count += 1
    avg = total / count   # WRONG! Prints a wrong partial answer every iteration
    print("avg so far:", avg)  # Only the LAST print is the real answer
````

````python
# ✅ Accumulator with a condition (only count/sum valid values)
temp_data = [-5, -2, 0, 1000, 2, -12, 44, 40, 30, -10]
min_val, max_val = -10, 40

running_total = 0
count = 0

for i in range(len(temp_data)):
    if temp_data[i] >= min_val and temp_data[i] <= max_val:  # only valid values
        running_total += temp_data[i]
        count += 1

avg = running_total / count   # AFTER loop
print("Average (valid only):", avg)

expected = (-5 + (-2) + 0 + 2 + 40 + 30 + (-10)) / 7
print("Expected:", expected)
````

## 8. Loop Pattern: Building Lists

A very common pattern: start with an empty list, then append items that meet a condition.

````python
# ✅ Pattern: start empty, append items that pass a condition
cut_off = 7.0
titles  = ["Umbrellas of Cherbourg", "M3gan", "Psycho", "Melania"]
ratings = [7.8, 6.3, 8.5, 1.3]

good_movies  = []   # start empty
ok_movies    = []
bad_movies   = []

for title, rating in zip(titles, ratings):
    if rating >= 7.0:
        good_movies.append(title)
    elif rating >= 6.0:
        ok_movies.append(title)
    else:
        bad_movies.append(title)
        print("Warning! Bad movie:", title)

print("Good:", good_movies)
print("OK:",   ok_movies)
print("Bad:",  bad_movies)
````

````python
# ✅ Building a list to store simulation results
# (alternative to pre-allocating with np.zeros)

xbar_list = []   # start empty

for i in range(500):
    sample = np.random.normal(size=30)
    xbar_list.append(sample.mean())   # add one result each iteration

print("Number of results stored:", len(xbar_list))
print("Mean of all xbars:", np.mean(xbar_list))
````

````python
# Two ways to add to a list — both are fine
my_list = []

# Method 1: .append()
my_list.append("hello")
my_list.append(42)
print(my_list)   # ['hello', 42]

# Method 2: concatenation with +
my_list = []
my_list = my_list + ["hello"]
my_list = my_list + [42]
print(my_list)   # ['hello', 42]
````

## 9. Loop Pattern: `continue` and `break`

| Statement | Effect |
|-----------|--------|
| `continue` | Skip the rest of THIS iteration; move to the NEXT one |
| `break` | Exit the entire loop immediately |

````python
# ✅ continue: skip non-positive values when summing
random_array = np.random.uniform(low=-5, high=10, size=20)

pos_sum = 0
for val in random_array:
    if val <= 0:
        continue       # skip this iteration entirely
    pos_sum += val     # only runs if val > 0

print("Sum of positive values:", pos_sum)

# Verify:
print("Check (numpy):", random_array[random_array > 0].sum())
````

````python
# ✅ continue: track position of non-numeric items
list_mixed = [1, 2, None, 4, "missing", 11, 2.2, None, "missing", 3]

list_sum = 0
position = 0

for value in list_mixed:
    if type(value) != int and type(value) != float:
        print("Warning: Non-numeric at position", position)
        position += 1
        continue        # skip the addition below
    list_sum += value
    position += 1

print("Sum of numeric values:", list_sum)
````

````python
# ✅ break: stop processing when bad sensor data is found
temp_data = [-5, -2, 0, 1000, 2, -12, 44, 40, 30, -10]
min_val, max_val = -10, 40

running_total = 0
count = 0

for i in range(len(temp_data)):
    if temp_data[i] < min_val or temp_data[i] > max_val:
        print("Warning: bad reading detected at position", i, "- stopping.")
        break          # exit loop immediately
    running_total += temp_data[i]
    count += 1

if count > 0:
    avg = running_total / count
    print("Average of readings before error:", avg)
````

## 10. Loop Pattern: Plotting Inside Loops

Two scenarios: multiple separate plots, or a grid of subplots.

````python
# ✅ Multiple separate plots (one plt.show() per loop iteration)
import pandas as pd

# Simulate a small dataset
np.random.seed(42)
df = pd.DataFrame({
    'mpg':          np.random.uniform(15, 45, 50),
    'displacement': np.random.uniform(100, 400, 50),
    'weight':       np.random.uniform(2000, 5000, 50),
    'acceleration': np.random.uniform(8, 24, 50)
})

x_variables = ['displacement', 'weight', 'acceleration']
y_var = 'mpg'

for x_var in x_variables:
    plt.scatter(x=df[x_var], y=df[y_var])
    plt.xlabel(x_var)
    plt.ylabel(y_var)
    plt.title(x_var + " vs " + y_var)
    plt.show()   # ← call show() INSIDE loop to get separate figures
````

````python
# ✅ Subplot grid (call show() AFTER the loop)
fig, subfig_array = plt.subplots(1, 3, figsize=(12, 4))

for i, x_var in enumerate(x_variables):   # enumerate gives (index, value)
    subfig_array[i].scatter(x=df[x_var], y=df[y_var])
    subfig_array[i].set_xlabel(x_var)
    subfig_array[i].set_ylabel(y_var)
    subfig_array[i].set_title(x_var + " vs " + y_var)

plt.tight_layout()
plt.show()   # ← call show() OUTSIDE the loop for a single combined figure
````

## 11. 🗺️ The Decision Guide — Which Loop Do I Use?

Ask yourself these questions in order:

```
Do I need to loop over a list?
│
├── YES: Do I need to MODIFY the list?
│   ├── YES → Type 2: for i in range(len(my_list)):
│   │
│   └── NO: Am I looping over TWO+ lists at the same time?
│       ├── YES → Type 4: for a, b in zip(list1, list2):
│       └── NO  → Type 1: for val in my_list:
│
└── NO: Am I repeating something N times (simulation, sum, etc.)?
    ├── YES → Type 3: for i in range(n):
    └── For each outer step, repeat inner steps → Type 5: Nested loops
```

````python
# Quick examples to test the decision guide

# Q: Print every movie title
# → Need values? YES. Modify list? NO. Two lists? NO.
# → Use Type 1: for val in list
movies = ["Rear Window", "Inception", "Psycho"]
for movie in movies:
    print(movie)
````

````python
# Q: Replace all negative values in a list with 0
# → Need values? YES. Modify list? YES.
# → Use Type 2: for i in range(len(...))
data = [3, -1, 5, -4, 2]
for i in range(len(data)):
    if data[i] < 0:
        data[i] = 0
print(data)
````

````python
# Q: Print "The movie X has rating Y" for each movie
# → Two lists? YES.
# → Use Type 4: for a, b in zip(...)
titles  = ["Rear Window", "Inception", "Psycho"]
ratings = [8.5, 8.8, 8.5]
for title, rating in zip(titles, ratings):
    print("The movie", title, "has rating", rating)
````

````python
# Q: Compute sum of integers 1 to 50
# → No list. Repeat 50 times.
# → Use Type 3: for i in range(n)
total = 0
for i in range(1, 51):
    total += i
print("Sum 1 to 50:", total)
````

````python
# Q: Run 1000 simulations for each of 3 different sample sizes
# → For each sample size, repeat 1000 times.
# → Use Type 5: Nested loops
for sample_sz in [10, 50, 200]:            # outer: each sample size
    results = []
    for _ in range(1000):                   # inner: 1000 simulations
        results.append(np.random.normal(size=sample_sz).mean())
    print(f"n={sample_sz}: std of xbar = {np.std(results):.4f}")
````

## 12. ⚠️ Common Mistakes — With Fixes

These are the most common errors on quizzes. Study each one carefully.

````python
# ❌ MISTAKE 1: Using values as indices
# Happens when you loop 'for i in my_list' then try 'my_list[i]'

my_list = [10, 20, 30, 40]

# WRONG:
# for i in my_list:          # i = 10, 20, 30, 40  (the VALUES)
#     print(my_list[i])      # ERROR: trying to use 10, 20, 30, 40 as indices!

# ✅ FIX: use range(len(...)) to get actual indices
for i in range(len(my_list)):   # i = 0, 1, 2, 3  (valid indices)
    print(my_list[i])
````

````python
# ❌ MISTAKE 2: Accumulator initialized inside the loop
data = [3, 7, 2, 9, 1]

# WRONG:
for val in data:
    total = 0           # resets to 0 EVERY iteration!
    total = total + val
print("Wrong:", total)  # Only has the last value

# ✅ FIX:
total = 0               # initialize BEFORE loop
for val in data:
    total = total + val
print("Correct:", total)
````

````python
# ❌ MISTAKE 3: Final computation inside the loop
data = [3, 7, 2, 9, 1]
total = 0
count = 0

# WRONG:
for val in data:
    total += val
    count += 1
    avg = total / count   # calculates a wrong intermediate average every iteration!
print("Last avg printed (happens to be correct, but wrong approach):", avg)

# ✅ FIX: compute after the loop
total = 0
count = 0
for val in data:
    total += val
    count += 1
avg = total / count   # AFTER loop
print("Correct avg:", avg)
````

````python
# ❌ MISTAKE 4: range(my_list) instead of range(len(my_list))
my_list = [10, 20, 30]

# WRONG:
# for i in range(my_list):      # TypeError! range() needs an integer, not a list
#     print(my_list[i])

# ✅ FIX:
for i in range(len(my_list)):   # len() gives the integer
    print(my_list[i])
````

````python
# ❌ MISTAKE 5: Using == instead of = to reset a variable
# This came up in the consecutive dice roll problem!

# WRONG: double3 == 0  is a COMPARISON (True/False), not an assignment!
# The variable double3 stays unchanged!

# ✅ FIX: use = to assign
double3 = 0
for roll in [1, 3, 3, 2, 4, 3, 3]:  # demo rolls
    if roll == 3:
        double3 += 1
    else:
        double3 = 0   # RESET with = (assignment)
    print(f"roll={roll}, double3={double3}")
````

````python
# ❌ MISTAKE 6: off-by-one in range for sum
# Computing sum 1 to n
n = 5

# WRONG: range(n) gives 0, 1, 2, 3, 4 — misses n and includes 0!
total = 0
for i in range(n):
    total += i
print("Wrong sum:", total)   # 0+1+2+3+4 = 10, but we want 1+2+3+4+5 = 15

# ✅ FIX: range(1, n+1) gives 1, 2, 3, 4, 5
total = 0
for i in range(1, n + 1):
    total += i
print("Correct sum:", total)
````

````python
np.random.seed(42)
random_array = np.random.uniform(low=-5, high=10, size=25)

# Your code here (use a for loop with continue)
````

````python
# SOLUTION
pos_sum = 0
for val in random_array:
    if val <= 0:
        continue
    pos_sum += val

print("Sum of positives:", pos_sum)
print("Check:", random_array[random_array > 0].sum())
````

Flag patients who are at risk of hypertension. A patient is at risk if their systolic blood pressure is **≥ 140 AND** their age is **≥ 65**. Build a list called `at_risk` of patient IDs who meet both criteria.

````python
patient_ids  = ["A1023", "A1044", "B2331", "B2440", "C3001",
                "C3120", "D4105", "D4119", "E5202", "E5307"]
bp_lst       = [138, 142, 152, 171, 155, 119, 144, 140, 132, 139]
age_lst      = [ 72,  60,  68,  65,  82,  34,  55,  70,  67,  75]

at_risk = []
bp_cutoff  = 140
age_cutoff = 65

# Your code here
````
