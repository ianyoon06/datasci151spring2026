## Basic Syntax

```python
# Loop over a range of numbers

i = [0,1,2,3,4]

for i in range(5):        # i = 0, 1, 2, 3, 4
    print(i)

for i in range(1, 6):     # i = 1, 2, 3, 4, 5
    print(i)

for i in range(1, n+1):   # i = 1, 2, ..., n  (inclusive of n)
    print(i)
```

## Looping Over a List

```python
my_list = [10, 20, 30, 40]

# By VALUE (cannot modify the list)
for val in my_list:
    print(val)             # val = 10, 20, 30, 40

# By INDEX (can modify the list)
for i in range(len(my_list)):
    print(my_list[i])      # i = 0, 1, 2, 3
```

## range() Key Facts
- `range(stop)` → 0 to stop-1
- `range(start, stop)` → start to stop-1 (**stop is exclusive**)
- `range(len(my_list))` → indices 0, 1, ..., len-1 (covers every element)

## Accumulator Pattern (Running Total)

```python
# Sum of 1 to n
total = 0                  # initialize BEFORE the loop
for i in range(1, n+1):
    total += i             # same as: total = total + i
print(total)
```

## Conditional Inside a Loop

```python
# Average of valid values only (ignore outliers)
data = [-5, -2, 0, 1000, 2, -12, 44, 40, 30, -10]
min_val, max_val = -10, 40

total = 0
count = 0
for i in range(len(data)):
    if data[i] >= min_val and data[i] <= max_val:
        total += data[i]
        count += 1

avg = total / count        # compute AFTER the loop
print(avg)
```

## Modifying a List In-Place

```python
# Replace outliers with None
for i in range(len(data)):
    if data[i] > max_val or data[i] < min_val:
        data[i] = None     # must use index to modify the list
```

## break — Exit Loop Early

```python
# Stop and warn on first bad value
total = 0
count = 0
for i in range(len(data)):
    if data[i] >= min_val and data[i] <= max_val:
        total += data[i]
        count += 1
    else:
        print("Warning: error detected in sensor.")
        break              # exits the loop immediately

avg = total / count
print(avg)
```

## Common Mistakes
| Mistake | Fix |
|---|---|
| `for i in my_list` then `my_list[i]` | Use `range(len(my_list))` for indices |
| Forgetting `n+1` in range | `range(1, n+1)` to include n |
| Computing average inside loop | Put `avg = total/count` **after** loop |
| `x == 0` instead of `x = 0` | `==` is comparison, `=` is assignment |
