### 1.2 Variables & Assignment Operator

## 2. Lists, Tuples & Strings

### 2.1 Lists

Lists store multiple objects. Created with **square brackets `[ ]`**. Elements separated by commas.

````python
movies  = ['Rear Window', 'Inception', 'Psycho']  # list of strings
numbers = [1, 2.5, -3, 0]                         # list of numbers
mixed   = [2, 'hello', True, None]                # mixed types allowed

print(len(movies))    # 3 — number of elements
print(movies[0])      # 'Rear Window'  — ZERO-indexed!
print(movies[1])      # 'Inception'
print(movies[-1])     # 'Psycho'       — last element
print(movies[-2])     # 'Inception'    — second to last
````

#### Modifying Lists
Lists are **MUTABLE** — you can change them after creation.

````python
movies[2] = 'The Wailing'              # change element at index 2
movies.append('Mulholland Drive')      # add to end
movies.sort()                          # sort in place
movies.reverse()                       # reverse in place
count = movies.count('The Wailing')    # count occurrences
print(movies)
````

#### List Concatenation (`+`)

````python
list_1 = [2, 3, 5]
list_2 = [11, 13]
list_3 = list_1 + list_2    # [2, 3, 5, 11, 13]
print(list_3)

# Growing a list inside a loop:
good_movies = []
good_movies = good_movies + ['Inception']   # ['Inception']
print(good_movies)
````

#### ⚠️ CRITICAL WARNING: List Assignment vs Copy

````python
a = [1, 2, 3]
b = a           # b is NOT a copy! Both point to the SAME list
a[0] = 99
print(b)        # [99, 2, 3]  <-- b also changed!

# To make a real copy:
c = a.copy()
a[0] = 0
print(c)        # [99, 2, 3]  <-- c is unaffected
````

### 2.2 Tuples

Tuples use **parentheses `( )`** and are **IMMUTABLE** (cannot be changed after creation).

````python
t = (5, 3, 2, 0)    # tuple with parentheses
t = 5, 3, 2, 0      # also a tuple! parentheses optional
print(t[0])         # access like a list: gives 5

# t[0] = 99         # ERROR! Tuples cannot be modified

# Single-element tuple MUST have a trailing comma:
one       = (1,)    # this IS a tuple
not_tuple = (1)     # this is just the integer 1!
print(type(one), type(not_tuple))
````

### 2.3 Strings

Strings are **IMMUTABLE** sequences of characters. You can index them like lists.

````python
name = 'Peter Sentz'
print(name[0])      # 'P'
print(name[6])      # 'S'
print(name[-1])     # 'z'
print(len(name))    # 11

# Concatenation with +
full = 'Peter' + ' ' + 'Sentz'    # 'Peter Sentz'
print(full)

# name[0] = 'p'   # ERROR! Strings are immutable
````

````python
# Tip: Use str() to convert numbers to strings for concatenation
age = 25
msg = 'I am ' + str(age) + ' years old.'   # must convert int to str
print(msg)
````

## 3. NumPy Arrays

### 3.1 Creating Arrays

````python
a = np.array([5.6, 2.9, 9.7, -10, 0, 3])   # from a list
z = np.zeros(5)                              # [0., 0., 0., 0., 0.]
o = np.ones(4)                               # [1., 1., 1., 1.]
r = np.linspace(0, 1, 5)                     # 5 evenly spaced from 0 to 1

print(a)
print(z)
print(r)
````

### 3.2 Array vs List: Key Difference

| Operation | List result | Array result |
|-----------|-------------|---------------|
| `a + b` | Concatenation: `[1,2]+[3,4]=[1,2,3,4]` | Element-wise: `[1,2]+[3,4]=[4,6]` |
| `a * 2` | Duplication: `[1,2]*2=[1,2,1,2]` | Scalar multiply: `[1,2]*2=[2,4]` |
| `a ** 2` | ERROR | Element-wise: `[1, 4, 9, ...]` |

````python
a = np.array([2, 3, 5, 11])
b = np.array([1, -4, 2, 0])

print(a + b)    # [3, -1, 7, 11]   — element-wise addition
print(a * 2)    # [4, 6, 10, 22]   — scalar multiply
print(a ** 2)   # [4, 9, 25, 121]  — element-wise power
print(a + 10)   # [12, 13, 15, 21] — broadcasting
````

### 3.3 Summary Statistics

````python
v = np.array([4.2, 1.2, 11.45, -2, 0, 9, 3.3])

# Method 1: numpy functions
print(np.mean(v))     # mean
print(np.std(v))      # standard deviation
print(np.median(v))   # median
print(np.max(v))      # maximum
print(np.min(v))      # minimum
print(np.size(v))     # number of elements

# Method 2: array methods
print(v.mean())       # same as np.mean(v)
print(v.std())        # same as np.std(v)
print(v.max())        # same as np.max(v)
print(v.min())        # same as np.min(v)
# NOTE: no .median() method — use np.median() instead
````

### 3.4 Copying Arrays

````python
x = np.array([1.2, 2.4, 4.8])
not_copy = x          # SAME object — changes affect both!
copy = np.copy(x)     # independent copy (use np.copy, not .copy())

x[0] = -99
print('not_copy:', not_copy)   # also changed!
print('copy:    ', copy)       # unaffected
````

### 3.5 Append

````python
z = np.array([5.1, 3.0, -2, 4.4])
z = np.append(z, 12)               # append one value — must reassign!
z = np.append(z, [13.1, 0.0])      # append multiple values
print(z)

# NOTE: np.append() returns a NEW array, does NOT modify in place
# Forgetting to reassign z is a common mistake!
````

## 4. Boolean Values & Comparison Operators

### 4.1 Boolean Objects

````python
# Only two boolean values:
print(type(True))    # <class 'bool'>   — capital T
print(type(False))   # <class 'bool'>   — capital F

# Don't confuse with strings!
print(type('True'))  # <class 'str'>    — this is a string!
````

### 4.2 Comparison Operators

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `3 < 1` | `False` |
| `>=` | Greater than or equal | `5 >= 5` | `True` |
| `<=` | Less than or equal | `4 <= 3` | `False` |
| `==` | Equal to (**NOT** assignment!) | `5*3 == 15` | `True` |
| `!=` | Not equal to | `5 != 6` | `True` |

> ⚠️ **WARNING:** `=` is assignment, `==` tests equality. Very common mistake!

````python
print(5 > 3)      # True
print(5 >= 5)     # True   — >= includes equal!
print(5 > 5)      # False  — strict greater than
print(5*3 == 15)  # True
print(5 != 6)     # True
````

### 4.3 Logical Operators

| Operator | Meaning | Example |
|----------|---------|--------|
| `not` | Negates a boolean | `not (age < 18)` → True if age ≥ 18 |
| `and` | Both must be True | `x > 0 and x < 10` |
| `or` | At least one must be True | `x < 0 or x > 100` |

````python
age = 22
print(age >= 18)          # True
print(not age < 18)       # True  — same thing, different phrasing

x = 5
print(x > 0 and x < 10)   # True  — both conditions met
print(x < 0 or x > 100)   # False — neither condition met
````

### 4.4 The `in` Keyword

````python
# Test if item belongs to a list, string, or array
months = ['June', 'July', 'August']
print('June' in months)      # True
print('April' in months)     # False

sentence = 'Federal Reserve'
print('Federal' in sentence) # True
print('federal' in sentence) # False — case-sensitive!

arr = np.array([1.0, 3.0, -17])
print(1 in arr)              # True  (1 == 1.0 in Python)
````

### 4.5 NumPy Array Comparisons

````python
# Returns a boolean array — one result per element
a = np.array([1, 2, 3])
b = np.array([1, 2, 4])
print(a == b)    # [True, True, False]
print(a >= b)    # [True, True, False]
````

## 5. If / Elif / Else Statements

### 5.1 Basic Structure

```python
if condition_1:
    # runs if condition_1 is True
elif condition_2:
    # runs if condition_1 False AND condition_2 True
elif condition_3:
    # runs if conditions 1 & 2 False AND condition_3 True
else:
    # runs if ALL conditions above are False
```

**Key rules:**
- Indentation is **MANDATORY** (4 spaces or 1 tab)
- Once a condition is `True`, Python executes that block and **skips the rest**
- Only the **FIRST** True condition executes — order matters!
- `else` and `elif` are optional

````python
# Grade classification example
final_grade = np.random.uniform(0, 100)
print("Grade:", round(final_grade, 1))

if final_grade >= 87:
    print('A')
elif final_grade >= 75:
    print('B')
elif final_grade >= 60:
    print('C')
elif final_grade >= 55:
    print('D')
else:
    print('F')
````

````python
# Piecewise function example
# y = 0 if x<=0, 0.5x if 0<x<=0.5, 0.5(1-x) if 0.5<x<=1, else 0

x = random.uniform(-0.5, 1.5)
print("x =", round(x, 3))

if x <= 0:
    y = 0
elif x <= 0.5:
    y = 0.5 * x
elif x <= 1:
    y = 0.5 * (1 - x)
else:
    y = 0

print("y =", y)
````

````python
# Nested if example
import random
first_word_choices = ["Introduction", "Intro", "Guide"]
selected = random.choices(first_word_choices, k=2)
title_py = selected[0] + " to Python"
title_r  = selected[1] + " to R"

keyword = "Introduction"

if keyword in title_py:
    if keyword in title_r:
        print('Keyword appears in both strings')
    else:
        print('Keyword appears only in title_py')
elif keyword in title_r:
    print('Keyword appears only in title_r')
else:
    print('Keyword appears in neither string')
````

## 6. For Loops

### 6.1 Basic Syntax

| Type | Syntax | Use when... |
|------|--------|-------------|
| By value | `for val in my_list:` | Just need the values, no modification |
| By index | `for i in range(len(my_list)):` | Need to modify list, or need the position |
| Count n times | `for i in range(n):` | Repeat something n times |
| Two lists | `for a, b in zip(list1, list2):` | Loop two+ lists simultaneously |

````python
# Loop by VALUE (cannot modify list in-place)
car_list = ["KIA", "Ferrari", "Ford", "Tesla"]
for car_name in car_list:
    print("Dear customer, we are writing about your " + car_name + " vehicle.")
````

````python
# Loop by INDEX (can modify list in-place)
numbers = [1, 2, 3, 4, 5]
for i in range(len(numbers)):
    numbers[i] = numbers[i] * 10
print(numbers)   # [10, 20, 30, 40, 50]
````

### 6.2 The `range()` Function

| Syntax | Values produced | Use case |
|--------|----------------|----------|
| `range(5)` | `0, 1, 2, 3, 4` | Loop 5 times |
| `range(1, 6)` | `1, 2, 3, 4, 5` | Loop from 1 to 5 (stop is **EXCLUSIVE**) |
| `range(1, n+1)` | `1, 2, ..., n` | Sum of first n integers |
| `range(len(lst))` | `0, 1, ..., len-1` | All valid indices of `lst` |

> ⚠️ `range(my_list)` causes a **TypeError**! You need `range(len(my_list))`.

````python
print(list(range(5)))        # [0, 1, 2, 3, 4]
print(list(range(1, 6)))     # [1, 2, 3, 4, 5]
print(list(range(1, 8)))     # [1, 2, 3, 4, 5, 6, 7]  (when n=7)
````

### 6.3 The `zip()` Function

````python
titles    = ["Umbrellas of Cherbourg", "M3gan", "Psycho"]
ratings   = [7.8, 6.3, 8.5]
years     = ["1964", "2022", "1960"]
directors = ["Jacques Demy", "Gerard Johnstone", "Alfred Hitchcock"]

# zip over 2 lists
for title, rating in zip(titles, ratings):
    print("The movie " + title + " has an IMDB rating of " + str(rating))

print()

# zip over 4 lists
for title, rating, year, director in zip(titles, ratings, years, directors):
    msg = "'" + title + "' (" + year + ") dir. " + director + " | " + str(rating)
    print(msg)

# ⚠️ zip stops at the SHORTEST list!
````

### 6.4 Accumulator Pattern

**Three golden rules:**
1. Initialize **BEFORE** the loop
2. Update **INSIDE** the loop  
3. Use the result **AFTER** the loop

````python
# ✅ CORRECT
n = 10
total = 0              # ① BEFORE
for i in range(1, n + 1):
    total = total + i  # ② INSIDE
print("Sum:", total)   # ③ AFTER
print("Check:", n*(n+1)//2)
````

````python
# ❌ WRONG — accumulator inside loop resets every iteration!
for i in range(1, 11):
    total = 0      # resets each time!
    total += i
print("Wrong total:", total)  # only has the last value (10)
````

### 6.5 Building Lists Inside Loops

````python
cut_off = 7.0
titles  = ["Umbrellas of Cherbourg", "M3gan", "Psycho", "Melania"]
ratings = [7.8, 6.3, 8.5, 1.3]

good_movies = []   # start empty
ok_movies   = []
bad_movies  = []

for title, rating in zip(titles, ratings):
    if rating >= 7.0:
        good_movies.append(title)
    elif rating >= 6.0:
        ok_movies.append(title)
    else:
        bad_movies.append(title)
        print("Warning! Bad movie:", title)

print("Good:", good_movies)
print("OK:  ", ok_movies)
print("Bad: ", bad_movies)
````

### 6.6 Plotting Inside Loops

````python
# Generate a small demo dataset
np.random.seed(42)
df_cars = pd.DataFrame({
    'mpg':          np.random.uniform(15, 45, 50),
    'displacement': np.random.uniform(100, 400, 50),
    'weight':       np.random.uniform(2000, 5000, 50),
    'acceleration': np.random.uniform(8, 24, 50)
})

y_var = 'mpg'
x_variables = ['displacement', 'weight', 'acceleration']

for x_var in x_variables:
    plt.scatter(x=df_cars[x_var], y=df_cars[y_var])
    plt.xlabel(x_var)
    plt.ylabel(y_var)
    plt.title(x_var + " vs " + y_var)
    plt.show()   # call show() INSIDE the loop for separate figures
````

### 6.7 `continue` and `break`

| Statement | Effect |
|-----------|--------|
| `continue` | Skip the rest of THIS iteration; go to the next one |
| `break` | Exit the loop **immediately** |

````python
# continue: skip non-positive values when summing
np.random.seed(0)
random_array = np.random.uniform(-5, 10, 20)

pos_sum = 0
for val in random_array:
    if val <= 0:
        continue       # skip to next iteration
    pos_sum += val

print("Sum of positives:", pos_sum)
````

````python
# break: stop processing when bad sensor data is found
temp_data = [-5, -2, 0, 1000, 2, -12, 44, 40, 30, -10]
min_val, max_val = -10, 40

running_total = 0
count = 0

for i in range(len(temp_data)):
    if temp_data[i] < min_val or temp_data[i] > max_val:
        print("Warning: bad reading at position", i, "— stopping.")
        break
    running_total += temp_data[i]
    count += 1

if count > 0:
    print("Average before error:", running_total / count)
````

### 6.8 Nested Loops

````python
# Nested loops + zip for subplot grid (from Lecture 9)
sample_sizes = [1, 10, 50, 100, 150, 200]
fig_rows     = [0, 0, 0, 1, 1, 1]
fig_cols     = [0, 1, 2, 0, 1, 2]

fig, subfig_array = plt.subplots(2, 3, figsize=(9, 6))

for sample_sz, row, col in zip(sample_sizes, fig_rows, fig_cols):  # outer
    vec_xbar = np.zeros(500)
    for i in range(500):                                            # inner
        vec_xbar[i] = np.random.uniform(-2, 2, size=sample_sz).mean()
    subfig_array[row, col].hist(vec_xbar)
    subfig_array[row, col].set_title("n = " + str(sample_sz))

plt.tight_layout()
plt.show()
````

### 6.9 Common For Loop Mistakes

| Mistake | Symptom | Fix |
|---------|---------|-----|
| `for i in my_list` then `my_list[i]` | `IndexError` (value used as index) | Use `range(len(my_list))` |
| `total = 0` inside loop | Total resets each iteration | Initialize **before** loop |
| `print(avg)` inside loop | Prints partial result | Put computation **after** loop |
| `range(my_list)` | `TypeError` | Use `range(len(my_list))` |
| `double3 == 0` instead of `= 0` | Variable not reset | Use `=` for assignment |

## 7. While Loops

```python
while condition:
    # body: runs as long as condition is True
    # MUST eventually make condition False, or loop runs forever!
```

````python
# Roll until you get a 6
num_rolls = 0
is_six = False
while not is_six:
    roll = random.choice([1, 2, 3, 4, 5, 6])
    num_rolls += 1
    if roll == 6:
        is_six = True
print("Rolls needed:", num_rolls)
````

````python
# Roll until two 3's total (not consecutive)
count = 0
num_rolls = 0
while count < 2:
    roll = np.random.choice([1, 2, 3, 4])
    num_rolls += 1
    if roll == 3:
        count += 1
print("Rolls to get two 3s:", num_rolls)
````

````python
# Roll until two CONSECUTIVE 3's
double3 = 0
num_rolls = 0
while double3 < 2:
    roll = np.random.choice([1, 2, 3, 4])
    num_rolls += 1
    if roll == 3:
        double3 += 1
    else:
        double3 = 0   # RESET — note: = not ==!
print("Rolls to get two consecutive 3s:", num_rolls)
````

### 9.3 Using zip for Subplots

````python
sample_sizes = [1, 10, 50, 100, 150, 200]
fig_rows     = [0, 0, 0, 1, 1, 1]
fig_cols     = [0, 1, 2, 0, 1, 2]

fig, subfig_array = plt.subplots(2, 3, figsize=(9, 6))

for sz, row, col in zip(sample_sizes, fig_rows, fig_cols):
    vec_xbar = np.zeros(2000)
    for i in range(2000):
        sample = np.random.normal(size=sz)
        vec_xbar[i] = sample.mean()
    subfig_array[row, col].hist(vec_xbar)
    subfig_array[row, col].set_title('n = ' + str(sz))

plt.tight_layout()
plt.show()
````

## 10. Random Variables with NumPy

### 10.1 Generating Random Samples

| Distribution | Function | Key parameters |
|---|---|---|
| Normal | `np.random.normal()` | `loc=mean, scale=std_dev, size=n` |
| Uniform | `np.random.uniform()` | `low=a, high=b, size=n` |
| Chi-squared | `np.random.chisquare()` | `df=degrees_of_freedom, size=n` |
| Choice (discrete) | `np.random.choice()` | `array_to_choose_from, size=n` |

````python
# Normal distribution (default: mean=0, std=1)
sample = np.random.normal(size=1000)                    # standard normal
sample = np.random.normal(loc=7, scale=5, size=1500)    # mean=7, sd=5

# Uniform distribution
sample = np.random.uniform(low=-2, high=2, size=100)

# Discrete choice
roll  = np.random.choice([1, 2, 3, 4, 5, 6])           # single die roll
rolls = np.random.choice([1, 2, 3, 4, 5, 6], size=10)  # 10 rolls
print(rolls)
````

### 10.2 Reproducibility with Seeds

````python
# Without seed: different results every run
print(np.random.normal(size=4))

# With seed: SAME results every run
np.random.seed(1234)   # set seed BEFORE generating
print(np.random.normal(size=4))   # same every time

np.random.seed(1234)   # reset same seed
print(np.random.normal(size=4))   # identical output
````

### 10.3 Simulating the Sample Mean (Central Limit Theorem)

````python
num_simulations = 2000
sample_size     = 100

vec_xbar = np.zeros(num_simulations)   # pre-allocate result array

for i in range(num_simulations):
    sample = np.random.uniform(low=-2, high=2, size=sample_size)
    vec_xbar[i] = sample.mean()        # store the sample mean

plt.hist(vec_xbar)
plt.title('Distribution of Sample Mean, n=' + str(sample_size))
plt.xlabel('Sample mean')
plt.ylabel('Frequency')
plt.show()
# As sample_size increases, the distribution approaches a bell curve!
````

### 12.1 Filter and Sort into Lists

````python
cut_off = 7.0
titles  = ['Umbrellas of Cherbourg', 'M3gan', 'Psycho', 'Melania']
ratings = [7.8, 6.3, 8.5, 1.3]

good_movies = []
ok_movies   = []
bad_movies  = []

for title, rating in zip(titles, ratings):
    if rating >= 7.0:
        good_movies.append(title)
    elif rating >= 6.0:
        ok_movies.append(title)
    else:
        bad_movies.append(title)
        print('Warning! Bad movie:', title)

print('Good:', good_movies)
print('OK:  ', ok_movies)
print('Bad: ', bad_movies)
````

### 12.2 Compute Vector from Formula

````python
coord_x = [-1, 0, 1, 2, 3]
coord_y = [-1, 0, 1, 2, 3]

# Approach 1: NumPy element-wise (PREFERRED)
z_array = 2*np.array(coord_x) + np.array(coord_y)**3 - 1
print('Approach 1:', z_array)

# Approach 2: For loop + append
z_list = []
for i in range(len(coord_x)):
    z_list.append(2*coord_x[i] + coord_y[i]**3 - 1)
z_array = np.array(z_list)
print('Approach 2:', z_array)

# Approach 3: Pre-allocate array
z_array = np.zeros(len(coord_x))
for i in range(len(coord_x)):
    z_array[i] = 2*coord_x[i] + coord_y[i]**3 - 1
print('Approach 3:', z_array)
````

### 12.3 Hypertension Risk Flag

````python
patient_id_lst  = ['A1023','A1044','B2331','B2440','C3001','C3120','D4105','D4119','E5202','E5307']
systolic_bp_lst = [138, 142, 152, 171, 155, 119, 144, 140, 132, 139]
age_lst         = [ 72,  60,  68,  65,  82,  34,  55,  70,  67,  75]

bp_cut_off  = 140
age_cut_off = 65
at_risk = []

for patient, bp, age in zip(patient_id_lst, systolic_bp_lst, age_lst):
    if bp >= bp_cut_off and age >= age_cut_off:
        at_risk.append(patient)

print('At-risk patients:', at_risk)
````
