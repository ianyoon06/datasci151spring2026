# 1. User-Defined Functions

### Why Functions?
A function is a **block of reusable code** that performs a specific task. Functions eliminate repetition and make code more manageable as it grows larger.

### Basic Syntax

```python
def my_function(param1, param2):
    # body: perform computations
    result = param1 + param2
    return result

# Call the function:
output = my_function(5, 10)
print(output)
```

### Key Rules
- **`def`** keyword starts the function definition, followed by the function name and parentheses
- **Parameters** (inputs) are listed inside the parentheses
- The body must be **indented** (just like loops and if/else)
- **`return`** specifies the output. Without `return`, the function returns **`None`**
- Parameter names inside the function are **arbitrary** — they don't need to match the caller's variable names

### Multiple Return Values
A function can return multiple values separated by commas. The result is a **tuple** that can be unpacked:

```python
import numpy as np

def quadratic_formula(a, b, c):
    discrim = b**2 - 4*a*c
    x_plus = (-b + np.sqrt(discrim)) / (2*a)
    x_minus = (-b - np.sqrt(discrim)) / (2*a)
    return x_plus, x_minus

# Unpack into two variables:
root1, root2 = quadratic_formula(3, -18, 15)
print("Root 1:", root1)
print("Root 2:", root2)

# Or capture as a single tuple:
output = quadratic_formula(3, -18, 15)
print("Type:", type(output))
print("First root:", output[0])
print("Second root:", output[1])
```

### Keyword Arguments When Calling

```python
# All valid:
quadratic_formula(3, -18, 15)
quadratic_formula(a=3, b=-18, c=15)
quadratic_formula(3, b=-18, c=15)

# INVALID -- positional arg after keyword arg:
# quadratic_formula(a=3, -18, 15)   # This would cause an error!
```

### Functions with No Return Value
Some functions perform an action (like printing or plotting) without returning a value. You can either omit the `return` statement entirely or write `return None`.

```python
import matplotlib.pyplot as plt

def create_histogram(vec_x, xlabel, title):
    plt.hist(x=vec_x)
    plt.xlabel(xlabel)
    plt.ylabel("Frequency")
    plt.title(title)
    plt.show()
    # no return statement needed
```

### Functions Calling Other Functions
You can build larger functions out of smaller ones:

```python
def F1(x, y, z):
    return x

def F2(x, y, z):
    return y**2 * z

def F3(x, y, z):
    return x * y * z**3

def F(x, y, z):
    return np.array([F1(x,y,z), F2(x,y,z), F3(x,y,z)])

print(F(-1, 2, -3))
```

# 3. Local vs. Global Variables

Variables created **inside** a function (that are not returned) are **local variables**. They exist only during the function call and are **not** accessible in your main workspace. Variables in your main workspace are **global variables**.

```python
import numpy as np

def quadratic_formula(a, b, c):
    discrim = b**2 - 4*a*c          # LOCAL variable
    x_plus = (-b + np.sqrt(discrim)) / (2*a)   # LOCAL
    x_minus = (-b - np.sqrt(discrim)) / (2*a)  # LOCAL
    return x_plus, x_minus

c1, c2, c3 = 3, -18, 15                # GLOBAL variables
z_plus, z_minus = quadratic_formula(c1, c2, c3)  # GLOBAL

print("z_plus:", z_plus)
print("z_minus:", z_minus)

# print(discrim)  # -> NameError! discrim is local to the function
```

> **Tip:** The variables `a`, `b`, `c` inside the function definition are also local. They receive copies of whatever values you pass in.

# 13. Merging DataFrames with `pd.merge()`

### Why Merge?
Datasets are often spread across multiple files. An **Entity Relationship Diagram (ERD)** shows how tables are connected via shared identifier columns. You can't just stack (concatenate) tables with different numbers of rows — instead you **merge** them on a shared key column.

### `pd.merge()` Syntax

| Argument | Description |
|----------|-------------|
| `left` | The first (primary) DataFrame |
| `right` | The second DataFrame |
| `how` | `"left"` = keep all rows from left, match from right. `"right"` = opposite |
| `on` | String: the column name both DataFrames share |

```python
# df_merged = pd.merge(
#     left  = results_short,
#     right = races_short,
#     how   = "left",
#     on    = "raceId"
# )
```

### How `"left"` vs `"right"` Works
- `how="left"`: left DF is primary, row order follows left. Extra columns from right are appended.
- `how="right"`: right DF is primary.

## Common Patterns from Lecture Exercises

### Compound Interest Function

```python
def invest(P, r, n, t):
    V = P * (1 + r/n) ** (n*t)
    return round(V, 2)

# Example: $100, 1% rate, compounded 20x/year, 10 years
print(invest(100, 0.01, 20, 10))
```

### Experiment Simulation Function

```python
def run_experiment(success_val):
    num_success = 0
    num_trials = 0
    while num_success < success_val:
        outcome = np.random.choice(
            ["Success", "Failure"], p=[0.25, 0.75]
        )
        num_trials = num_trials + 1
        if outcome == "Success":
            num_success = num_success + 1
    return num_trials

print(run_experiment(3))
print(run_experiment(5))
```

### Consecutive Successes Variant

```python
def run_consecutive(success_val):
    num_success = 0
    num_trials = 0
    while num_success < success_val:
        outcome = np.random.choice(
            ["Success", "Failure"], p=[0.8, 0.2]
        )
        num_trials = num_trials + 1
        if outcome == "Success":
            num_success = num_success + 1
        else:
            num_success = 0   # RESET on failure
    return num_trials

print(run_consecutive(3))
print(run_consecutive(7))
```

# 15. Practice Q1 — Looping Over Filenames with `os.listdir`

- Using `os.listdir()` to programmatically get file names from a folder
- Building file paths by concatenating the folder path + filename
- Using a for loop to read multiple CSVs and collect them into a list
- Using `len()` and `.shape` to inspect the results

### Key Idea
`os.listdir("some/folder")` returns a **list of file name strings** (just the names, NOT the full paths). You must prepend the folder path when passing each file name to `pd.read_csv()`.

```python
import os
import pandas as pd

# Step 1: Get file names
folder_path = "quiz_data/country_emissions"
file_names = os.listdir(folder_path)
print("Files found:", file_names)

# Step 2: Empty list
df_list = []

# Step 3: Loop, read each CSV, append
for file_name in file_names:
    full_path = folder_path + "/" + file_name
    df = pd.read_csv(full_path)
    df_list.append(df)

# Step 4: Print results
print("Number of DataFrames:", len(df_list))
print("Shape of first DataFrame:", df_list[0].shape)
```

### Common Mistakes to Avoid
- **Forgetting to build the full path**: `pd.read_csv(file_name)` will fail because `file_name` is just `"germany_emission.csv"`, not `"quiz_data/country_emissions/germany_emission.csv"`
- **Hardcoding file names**: The point of `os.listdir` is to avoid manually typing each file name
- Using `df_list.append(full_path)` instead of `df_list.append(df)` — you want the **DataFrame**, not the string

# 16. Practice Q2 — Alternate Quadratic Formula

### Concept Being Tested
- Writing a custom function with `def`, parameters, and `return`
- Translating a math formula into Python code
- Testing your function with a known answer

### The Formula
The alternate quadratic formula is:

$$x = \frac{2c}{-b \mp \sqrt{b^2 - 4ac}}$$

The $\mp$ means one solution uses **minus** and the other uses **plus**:
- $x_1 = \frac{2c}{-b - \sqrt{b^2 - 4ac}}$
- $x_2 = \frac{2c}{-b + \sqrt{b^2 - 4ac}}$

### Test Case
$3(x+1)(x-2) = 0$ expands to $3x^2 - 3x - 6 = 0$, so $a=3, b=-3, c=-6$. Roots should be $-1$ and $2$.

```python
import numpy as np

def alt_quad_formula(a, b, c):
    discrim = np.sqrt(b**2 - 4*a*c)
    x1 = (2*c) / (-b - discrim)
    x2 = (2*c) / (-b + discrim)
    return x1, x2

# Test: 3(x+1)(x-2) = 3x^2 - 3x - 6
# a = 3, b = -3, c = -6
root1, root2 = alt_quad_formula(3, -3, -6)
print("Root 1:", root1)   # should be -1 or 2
print("Root 2:", root2)   # should be 2 or -1
```

### Common Mistakes to Avoid
- Forgetting that the formula has $-b$ (not $+b$) in the denominator
- The $\mp$ means one root uses minus, the other uses plus — it's the **opposite** of the standard $\pm$
- Expanding the polynomial incorrectly: $3(x+1)(x-2) = 3(x^2 - x - 2) = 3x^2 - 3x - 6$

### Common Mistakes to Avoid
- Using `np.random.choice` instead of `np.random.uniform` — the question specifically asks for uniform
- Forgetting to set `size=n` — without it you get a single number, not an array
- Using `>` instead of `>=` for groupB — you'd lose rows where `random_var` is exactly 0.4
- Not storing the random numbers as a new column before querying
