### 1. Function Anatomy & The Four Types

Every function has the same skeleton:

```python
def function_name(param1, param2):    # HEADER
    # BODY (indented)
    result = param1 + param2
    return result                      # RETURN
```

| Component | Example |
|-----------|---------|
| `def` keyword | `def quadratic_formula(...)` |
| Function name | `quadratic_formula` |
| Parameters | `(a, b, c)` |
| Body (indented) | `discrim = b**2 - 4*a*c` |
| `return` | `return x_plus, x_minus` |

### Type A: Compute and Return a Value

```python
def invest(P, r, n, t):
    V = P * (1 + r/n) ** (n*t)
    return round(V, 2)

def alt_quad_formula(a, b, c):
    import numpy as np
    discrim = np.sqrt(b**2 - 4*a*c)
    x1 = (2*c) / (-b - discrim)
    x2 = (2*c) / (-b + discrim)
    return x1, x2     # returns a TUPLE

def check_voter_eligibility(age):
    return age >= 18  # returns True or False directly
```

### Type D: Simulation / While Loop

```python
import numpy as np

def run_consecutive(success_val):
    num_success = 0
    num_trials = 0
    while num_success < success_val:
        outcome = np.random.choice(["Success","Failure"], p=[0.8,0.2])
        num_trials += 1
        if outcome == "Success":
            num_success += 1
        else:
            num_success = 0   # RESET on failure
    return num_trials
```

### What Is a Column?
A DataFrame is a table. Each column has a **name** and holds one type of data. When you read a CSV, the first row of the file automatically becomes the column names.

### Single vs Multiple Column Extraction

| Code | What You Get | Type |
|------|--------------|------|
| `df["mpg"]` | One column of data | **Series** |
| `df[["mpg"]]` | One column wrapped in a table | **DataFrame** |

The double brackets are NOT special syntax — the inner brackets define a Python **list**, the outer brackets index the DataFrame.

Regular bracket extraction handles spaces and parentheses fine:

```python
# df["Height(in cm)"]      # works perfectly
# df["miles per gallon"]   # also fine
```

But inside `.query()`, you need **backticks**:

```python
# This FAILS:
# df.query("Height(in cm) > 175")

# This WORKS:
# df.query("`Height(in cm)` > 175")
```

### 13. Dictionary Basics

A **dictionary** is a Python container that stores data as **key-value pairs**. Unlike a list (which uses integer positions 0, 1, 2…), a dictionary uses any **key** you choose — strings, integers, etc.

### Why dictionaries matter
- They let you look up values by a meaningful name instead of remembering positions
- They're the easiest way to build small DataFrames for testing
- They appear inside `pd.DataFrame()`, `.replace()`, and `.agg()`

### The Problem Dictionaries Solve

```python
# With lists, you need TWO of them and have to track positions:
student_roster = ["Nancy", "Bill", "Elaine", "Boris", "Assata"]
student_grades = [92.8, 87.3, 95.4, 82.3, 98.7]

# To update Boris's grade, you have to find his position first:
b_index = student_roster.index("Boris")
student_grades[b_index] = 85
print(student_grades)
```

### With a Dictionary, You Use Names Directly

```python
grades = {
    "Nancy":  92.8,
    "Bill":   87.3,
    "Elaine": 95.4,
    "Boris":  82.3,
    "Assata": 98.7
}

# Look up by name
print(grades["Boris"])    # 82.3

# Update by name
grades["Boris"] = 85
print(grades)
```

### Keys and Values

```python
print(grades.keys())     # dict_keys(['Nancy', 'Bill', ...])
print(grades.values())   # dict_values([92.8, 87.3, ...])

# Loop over keys
for key in grades.keys():
    name = key
    score = grades[key]
    print("Grade for", name, ":", score)
```

### Dictionary Syntax Rules

- Use **curly braces** `{}` (NOT square brackets — those are lists)
- Each entry is `key: value`, separated by a colon
- Entries are separated by commas
- Keys are usually strings or integers
- Values can be ANY object: strings, numbers, lists, even other dictionaries

```python
example_dict = {
    0: "QTM-151",
    "favorite_movie": "Evil Dead (2013)",
    12: 35.2,
    "my_list": [-8, 0.2, 9.1],
    "nested_dict": grades   # a dict inside a dict!
}

print(example_dict[0])
print(example_dict["favorite_movie"])
print(example_dict["nested_dict"])
```

### 21. Three Ways to Recode

**Recoding** means transforming an existing variable into a new set of values that are more useful for analysis.

### Common examples
- `age` (integer) → `"child"`, `"teen"`, `"adult"`, `"senior"`
- `lat` (float, geographic latitude) → `"North"` or `"South"`
- `mpg` (float) → `"Low"`, `"Medium"`, `"High"` mileage

You can recode in three ways. Let's use the example: convert latitude into hemisphere where `lat ≤ 0` is `"South"` and `lat > 0` is `"North"`.

### Method 1: Manual For Loop

```python
nrows, ncols = circuits.shape
col_no = 5    # column 5 is "lat"
hemisphere = []

for k in range(nrows):
    lat_val = circuits.iloc[k, col_no]
    if lat_val <= 0:
        hemisphere.append("South")
    else:
        hemisphere.append("North")

circuits["hemisphere"] = hemisphere
```

This works but is **verbose** and **slow** on large datasets. It's the most explicit option.

### Method 2: `.apply()` with a Function

```python
def check_hemisphere(lat):
    if lat <= 0:
        return "South"
    else:
        return "North"

circuits["hemisphere"] = circuits["lat"].apply(check_hemisphere)
```

Cleaner. Note the function takes ONE value (one row's `lat`) and returns ONE label.

> ⚠️ **Common bug:** make sure the function uses its parameter name (`lat`) inside the body, not some other variable. In the lecture notebook there's an example where `check_hemisphere(lat)` accidentally references `lat_val` from outside the function — that's a bug!

### 22. `pd.cut()` Deep Dive

`pd.cut()` is the **best** way to recode a numeric variable into bins. It's purpose-built for this task.

```python
lat_bins = [-90, 0, 90]              # the cutoff points
lat_labels = ["South", "North"]      # one fewer label than bins

circuits["hemisphere"] = pd.cut(
    circuits["lat"],
    bins = lat_bins,
    right = True,        # right endpoint INCLUDED
    labels = lat_labels
)
```

### Understanding `pd.cut()` Arguments

| Argument | What It Means |
|----------|---------------|
| First arg | The Series to bin (e.g., `df["lat"]`) |
| `bins` | List of cutoff numbers — defines the boundaries |
| `labels` | List of labels — must have **one fewer** label than bin edges |
| `right` | If `True`, the right endpoint is included (≤). If `False`, only the left (>) |

### How `bins` and `labels` Match Up

With `bins = [-90, 0, 90]` and `labels = ["South", "North"]` and `right=True`:

- Interval 1: `-90 < lat ≤ 0` → labeled `"South"`
- Interval 2: `0 < lat ≤ 90` → labeled `"North"`

3 bin edges create 2 intervals, so you need 2 labels. **Always one fewer label than bin edges.**

### Using `np.inf` for Open-Ended Bins

For robustness (in case data falls outside your expected range), use `-np.inf` and `np.inf` as the outer bin edges.

```python
lat_bins = [-np.inf, 0, np.inf]
lat_labels = ["South", "North"]

circuits["hemisphere"] = pd.cut(
    circuits["lat"],
    bins = lat_bins,
    right = True,
    labels = lat_labels
)
# Now any lat ≤ 0 becomes South, any lat > 0 becomes North
```

### Three-Bin Example: Mileage Categories

```python
# Recode mpg into Low / Medium / High
# - Low if mpg ≤ 20
# - Medium if 20 < mpg ≤ 34
# - High if mpg > 34

bins = [-np.inf, 20, 34, np.inf]
labels = ["Low mileage", "Medium mileage", "High mileage"]

# df["mileage"] = pd.cut(df["mpg"], bins=bins, right=True, labels=labels)
```

### Key points
- 4 bin edges → 3 intervals → 3 labels
- Labels must be in the **same order** as the bins they correspond to
- `right=True` makes the boundaries `≤` (inclusive on the right)
- Values **outside** the bin range become `NaN` (so use `-np.inf`/`np.inf` for safety)

### Equivalent with `.apply()`

For comparison, here's the same recoding done with `.apply()` (also valid, just more verbose):

```python
def check_mileage(mpg):
    if mpg <= 20:
        return "Low mileage"
    elif mpg <= 34:
        return "Medium mileage"
    elif mpg > 34:
        return "High mileage"
    else:
        return np.nan   # handle missing values!

# df["mileage"] = df["mpg"].apply(check_mileage)
```

> ✅ **On the quiz**, both `pd.cut` and `.apply` approaches are likely acceptable. `pd.cut` is shorter and cleaner; `.apply` is more flexible if you need complex logic.

### 23. `.groupby()` Basics

**Grouping** = splitting your data into chunks based on a column's unique values, then computing something (like mean) for each chunk separately.

### Why group?
Computing `df["points"].mean()` gives you the average points across the **entire** dataset — usually not what you want. What you actually care about is "average points **per driver**" or "average points **per region**." That's grouping.

### 27. `plt.subplots()` for Side-by-Side Plots

When you want **multiple plots in one figure** (side by side or in a grid), use `plt.subplots()`.

### The basic call

```python
fig, subplt = plt.subplots(
    nrows = 1,         # 1 row of plots
    ncols = 2,         # 2 columns of plots
    figsize = (10, 4)  # overall figure size in inches
)
```

### What `plt.subplots()` returns

It returns **two** things:
1. `fig`: the overall figure object (rarely used directly)
2. `subplt`: an **array** of axes objects, one per subplot

For `ncols=2`, `subplt` has 2 elements: `subplt[0]` (left) and `subplt[1]` (right). For a grid like `nrows=2, ncols=2`, `subplt` is 2D.

### Plotting on Each Subplot

Instead of `plt.scatter()`, you call `.scatter()` on the specific subplot. The labeling functions also have different names:

```python
groupA = df_fifa.query("Group == 'A'")
groupB = df_fifa.query("Group == 'B'")

fig, subplt = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

# Left subplot — Group A
subplt[0].scatter(x=groupA["Age"], y=groupA["Overall"],
                  color="red", edgecolor="black")
subplt[0].set_xlabel("Age")
subplt[0].set_ylabel("Overall")
subplt[0].set_title("Data from Group A")

# Right subplot — Group B
subplt[1].scatter(x=groupB["Age"], y=groupB["Overall"],
                  color="blue", edgecolor="black")
subplt[1].set_xlabel("Age")
subplt[1].set_ylabel("Overall")
subplt[1].set_title("Data from Group B")

plt.tight_layout()   # prevents labels from overlapping
plt.show()
```

### The Critical Naming Differences

| Single plot (`plt.`) | Subplot (`subplt[i].`) |
|----------------------|------------------------|
| `plt.scatter(...)` | `subplt[0].scatter(...)` |
| `plt.hist(...)` | `subplt[0].hist(...)` |
| `plt.xlabel(...)` | `subplt[0].set_xlabel(...)` |
| `plt.ylabel(...)` | `subplt[0].set_ylabel(...)` |
| `plt.title(...)` | `subplt[0].set_title(...)` |
| `plt.legend(...)` | `subplt[0].legend(...)` |

> ⚠️ The big one to remember: it's `set_xlabel`, `set_ylabel`, `set_title` (with `set_`), NOT just `xlabel`, `ylabel`, `title`.

### When to Use Subplots

- Comparing two groups side by side (like Practice Quiz Q2)
- Showing multiple views of the same data
- Creating dashboards with several charts

For overlapping plots on a **single** axis, you don't need subplots — just call `plt.scatter()` (or `plt.hist()`) multiple times before `plt.show()`.

### 30. Indexing Into Return Values

When a function returns multiple values, Python packs them into a **tuple**.

```python
def alt_quad_formula(a, b, c):
    discrim = np.sqrt(b**2 - 4*a*c)
    return (2*c)/(-b - discrim), (2*c)/(-b + discrim)

# Choice 1: Unpack into separate variables
root1, root2 = alt_quad_formula(3, -3, -6)

# Choice 2: Store as tuple, index into it
output = alt_quad_formula(3, -3, -6)
print(output[0])    # first root
print(output[1])    # second root

# Ignoring a value with _
_, root2 = alt_quad_formula(3, -3, -6)
```

### 36. Q6 — Recoding Two Ways

### Concept Being Tested
- Reading CSV
- Recoding a numeric variable into a categorical one
- Knowing **both** the `.apply()` method AND the `pd.cut()` method

```python
df_cars = pd.read_csv("quiz_data/features.csv")

# Method 1: Using .apply()
def check_mileage(mpg):
    if mpg <= 20:
        return "Low mileage"
    elif mpg <= 34:
        return "Medium mileage"
    elif mpg > 34:
        return "High mileage"
    else:
        return np.nan   # handles missing values

df_cars["mileage_1"] = df_cars["mpg"].apply(check_mileage)

# Method 2: Using pd.cut()
bins = [-np.inf, 20, 34, np.inf]
labels = ["Low mileage", "Medium mileage", "High mileage"]

df_cars["mileage_2"] = pd.cut(
    x = df_cars["mpg"],
    bins = bins,
    right = True,
    labels = labels
)

# Verify the two methods produce the same result
check = list(df_cars["mileage_1"]) == list(df_cars["mileage_2"])
print(check)   # True
```

### Key things to notice
- Both methods give the same result, but `pd.cut` is shorter
- The `.apply()` solution needs an `else: return np.nan` to handle potential missing values
- `pd.cut` automatically returns NaN for values outside the bins, so using `-np.inf` and `np.inf` ensures everything is captured
- The order of `bins` and `labels` matters and they must align: `bins[i]` to `bins[i+1]` gets `labels[i]`
