# 1. Function Anatomy & The Four Types

````python
def function_name(param1, param2):    # HEADER
    # BODY (indented)
    result = param1 + param2
    return result                      # RETURN
````

| Component | Example |
|-----------|---------|
| `def` keyword | `def quadratic_formula(...)` |
| Function name | `quadratic_formula` |
| Parameters | `(a, b, c)` |
| Body (indented) | `discrim = b**2 - 4*a*c` |
| `return` | `return x_plus, x_minus` |

## Type A: Compute and Return

````python
def invest(P, r, n, t):
    V = P * (1 + r/n) ** (n*t)
    return round(V, 2)

def alt_quad_formula(a, b, c):
    discrim = np.sqrt(b**2 - 4*a*c)
    return (2*c)/(-b - discrim), (2*c)/(-b + discrim)

def check_voter_eligibility(age):
    return age >= 18

print(invest(1000, 0.05, 12, 10))
print(alt_quad_formula(3, -3, -6))
print(check_voter_eligibility(17))
````

## Type C: Classify (for `.apply()`)

## Type D: Simulation / While Loop

````python
def run_consecutive(success_val):
    num_success = 0
    num_trials = 0
    while num_success < success_val:
        outcome = np.random.choice(["Success","Failure"], p=[0.8,0.2])
        num_trials += 1
        if outcome == "Success":
            num_success += 1
        else:
            num_success = 0
    return num_trials
````

# 17. String Methods (`.str.isnumeric()`)

````python
# Returns True/False for each element
# Only works on object (string) columns!
print(messy["age"].str.isnumeric())
````

### What you should see
```
0     True   ← "21"
1     True   ← "19"
2    False   ← "\N"
3     True   ← "22"
4    False   ← "twenty"
5     True   ← "20"
```

### Find the bad rows

````python
bad_rows = messy.query("age.str.isnumeric() == False")
print(bad_rows)
````

> ⚠️ `.str.isnumeric()` is **strict**. Negative numbers like `"-7"` and decimals like `"3.14"` are also considered non-numeric because of the `-` and `.` characters.

---
# 18. `.replace()` for Cleaning

````python
# Step 1: Find the unique bad values
bad_values = pd.unique(bad_rows["age"])
print("Bad values:", bad_values)

# Step 2: Decide what to replace each with (same-length list!)
replacements = [np.nan, 20]   # \N -> NaN, "twenty" -> 20

# Step 3: Replace and overwrite the column
messy["age"] = messy["age"].replace(bad_values, replacements)
print(messy)
````

> ⚠️ The two lists must be the **same length**. Each bad value at position `i` is replaced with the value at position `i` in the second list.

---
# 19. `pd.to_numeric()` and `errors="coerce"`

````python
# Even after replacing, the dtype is still 'object'
print("Before to_numeric:")
print(messy.dtypes)
print()

# Convert age column to numeric
messy["age"] = pd.to_numeric(messy["age"])
print("After to_numeric:")
print(messy.dtypes)
print()

# Now we can do math!
print("Mean age:", messy["age"].mean())
````

### The shortcut: `errors="coerce"`

````python
# Automatically converts ANY non-numeric value to NaN
# Skips the manual .replace() step entirely

messy_v2 = pd.DataFrame({
    "name": ["Alice", "Bob", "Carol", "Dan"],
    "age":  ["21", "19", "\\N", "twenty"]
})

messy_v2["age"] = pd.to_numeric(messy_v2["age"], errors="coerce")
print(messy_v2)
print()
print(messy_v2.dtypes)
````

### When to use which approach
- **`.replace()` + `pd.to_numeric()`**: When some bad values have a meaningful replacement (like `"twenty"` → `20`)
- **`pd.to_numeric(errors="coerce")`**: When ALL bad values should just become NaN

# ⭐ Part E — Recoding Variables (with Worked Examples)
---

## Setup: Build a Toy DataFrame for Recoding

````python
scores = pd.DataFrame({
    "name":  ["Alice", "Bob", "Carol", "Dan", "Eve", "Frank", "Grace", "Henry"],
    "score": [95, 72, 88, 55, 41, 79, 65, 92]
})
print(scores)
````

# 21. Three Ways to Recode

**Goal:** convert the numeric `score` into a letter grade:
- `score >= 90` → "A"
- `80 <= score < 90` → "B"
- `70 <= score < 80` → "C"
- `score < 70` → "F"

## Method 1: Manual For Loop

````python
nrows = len(scores)
letter_grades = []

for k in range(nrows):
    s = scores.iloc[k, 1]      # column 1 is 'score'
    if s >= 90:
        letter_grades.append("A")
    elif s >= 80:
        letter_grades.append("B")
    elif s >= 70:
        letter_grades.append("C")
    else:
        letter_grades.append("F")

scores["grade_loop"] = letter_grades
print(scores)
````

## Method 2: `.apply()` with a Function

````python
def to_letter(s):
    if s >= 90:
        return "A"
    elif s >= 80:
        return "B"
    elif s >= 70:
        return "C"
    else:
        return "F"

# Pass the function NAME (no parentheses!) to .apply()
scores["grade_apply"] = scores["score"].apply(to_letter)
print(scores)
````

> ⚠️ The function `to_letter` takes a **single value**, not the whole Series. Pandas calls it once per row for you.

---
# 22. `pd.cut()` Deep Dive

`pd.cut()` is the cleanest way to recode a numeric variable into bins.

## Method 3: `pd.cut()` (the cleanest)

````python
# Define the cut points (bin edges)
bins = [-np.inf, 70, 80, 90, np.inf]

# Define the labels — ONE FEWER label than bin edges!
# 5 bin edges → 4 intervals → 4 labels
labels = ["F", "C", "B", "A"]

scores["grade_cut"] = pd.cut(
    scores["score"],
    bins = bins,
    right = False,    # left endpoint INCLUDED, right EXCLUDED
    labels = labels
)
print(scores)
````

### Understanding the bins

With `bins = [-np.inf, 70, 80, 90, np.inf]` and `right=False`:

| Interval | Range | Label |
|----------|-------|-------|
| 1 | `-inf <= score < 70` | F |
| 2 | `70 <= score < 80` | C |
| 3 | `80 <= score < 90` | B |
| 4 | `90 <= score < inf` | A |

**Always one fewer label than bin edges.** Here we have 5 bin edges and 4 labels.

## Comparing `right=True` vs `right=False`

````python
# right=True (default): right endpoint INCLUDED — bin is (a, b]
# right=False: right endpoint EXCLUDED — bin is [a, b)

# Test with values exactly at the boundaries
test_data = pd.DataFrame({"x": [69, 70, 71, 79, 80, 81, 89, 90, 91]})

test_data["with_right_True"]  = pd.cut(test_data["x"],
                                       bins=[-np.inf, 70, 80, 90, np.inf],
                                       right=True,
                                       labels=["F", "C", "B", "A"])

test_data["with_right_False"] = pd.cut(test_data["x"],
                                       bins=[-np.inf, 70, 80, 90, np.inf],
                                       right=False,
                                       labels=["F", "C", "B", "A"])

print(test_data)
````

Notice how a score of exactly 70 gets labeled differently:
- `right=True`: 70 falls into the **F** bin (`-inf < x ≤ 70`)
- `right=False`: 70 falls into the **C** bin (`70 ≤ x < 80`)

**Read the question carefully** to know which one to use!

## Why use `-np.inf` and `np.inf`?

````python
# Without infinity, values outside the bin range become NaN
unsafe = pd.cut([95, 50, 105],
                bins=[60, 70, 80, 90, 100],
                labels=["D", "C", "B", "A"])
print("Unsafe (50 and 105 become NaN):")
print(unsafe)
print()

# With infinity, every value gets a label
safe = pd.cut([95, 50, 105],
              bins=[-np.inf, 70, 80, 90, np.inf],
              labels=["F", "C", "B", "A"])
print("Safe (everything labeled):")
print(safe)
````

# ⭐ Part F — Grouping & Aggregating (with Worked Examples)
---

## Setup: Build a Sales Toy DataFrame

````python
sales = pd.DataFrame({
    "salesperson": ["Alice", "Bob", "Alice", "Carol", "Bob", "Alice", "Carol", "Bob", "Alice"],
    "category":    ["Books", "Books", "Toys", "Books", "Toys", "Toys", "Toys", "Books", "Books"],
    "amount":      [25, 40, 15, 60, 30, 22, 18, 50, 35]
})
print(sales)
````

# 23. `.groupby()` Basics

## Step 1: Group by salesperson, then take mean

````python
# Without grouping: average across ALL rows
print("Overall mean amount:", sales["amount"].mean())
print()

# With grouping: average PER salesperson
by_person = sales.groupby(by="salesperson")["amount"].mean()
print("Mean amount per salesperson:")
print(by_person)
````

## Other aggregation methods

````python
print("SUM per salesperson:")
print(sales.groupby("salesperson")["amount"].sum())
print()

print("COUNT per salesperson:")
print(sales.groupby("salesperson")["amount"].count())
print()

print("MAX per salesperson:")
print(sales.groupby("salesperson")["amount"].max())
````

## Sort the results to find top performers

````python
top_sellers = sales.groupby("salesperson")["amount"].sum().sort_values(ascending=False)
print(top_sellers)
print()

print("Top 2:")
print(top_sellers.iloc[:2])
````

## Grouping by Multiple Columns

````python
# For each (salesperson, category) combination
by_person_category = sales.groupby(by=["salesperson", "category"])["amount"].sum()
print(by_person_category)
````

---
# 24. `.agg()` for Multiple Statistics

## Simple `.agg()` on a Series

````python
# Pass a list of statistic NAMES (as strings)
stats = sales["amount"].agg(["mean", "std", "min", "max", "sum"])
print(stats)
````

## `.agg()` with Custom Names (Tuple Syntax)

````python
# Each line: new_name = (column_name, statistic_name)
summary = sales.agg(
    avg_sale   = ('amount', 'mean'),
    total_rev  = ('amount', 'sum'),
    biggest    = ('amount', 'max'),
    smallest   = ('amount', 'min'),
    n_sales    = ('amount', 'count')
)
print(summary)
````

## The Power Combo: Group + Agg

````python
salesperson_summary = sales.groupby("salesperson").agg(
    total_revenue = ('amount', 'sum'),
    avg_sale      = ('amount', 'mean'),
    biggest_sale  = ('amount', 'max'),
    num_sales     = ('amount', 'count')
)
print(salesperson_summary)
````

### Notice
- The column you reference (`'amount'`) goes in the first slot of the tuple
- The statistic name (`'sum'`, `'mean'`, etc.) goes in the second slot
- The variable name on the left (`total_revenue`, `avg_sale`...) becomes the column name in the result

---
# 25. Chaining: query + groupby + agg

````python
# Filter to Books only, then group by person, then aggregate
books_summary = (sales
                 .query("category == 'Books'")
                 .groupby("salesperson")
                 .agg(
                     book_revenue = ('amount', 'sum'),
                     avg_book     = ('amount', 'mean')
                 ))
print(books_summary)
````

### Reading the chain top-to-bottom
1. Start with `sales`
2. `.query("category == 'Books'")` filters to only Book transactions
3. `.groupby("salesperson")` groups what's left by salesperson
4. `.agg(...)` computes book_revenue and avg_book for each group

The outer parentheses around the whole expression let you split it across multiple lines.

## Quick alternative: chain to a single statistic

````python
# Average book sale per person — no .agg() needed for one stat
avg_book = sales.query("category == 'Books'").groupby("salesperson")["amount"].mean()
print(avg_book)
````

---
# 26. Aggregating + Merging Back

````python
# Step 1: Create per-person stats (one row per person)
person_stats = sales.groupby("salesperson").agg(
    avg_sale = ('amount', 'mean')
)
print("Per-person stats:")
print(person_stats)
print()

# Step 2: Merge those stats back into the original sales data
sales_with_stats = pd.merge(
    sales,
    person_stats,
    on = "salesperson",
    how = "left"
)
print("Each row labeled with that person's avg_sale:")
print(sales_with_stats)
````

Now every row has both the individual sale `amount` AND that salesperson's overall `avg_sale`. You could use this to compute deviations: `sales_with_stats["amount"] - sales_with_stats["avg_sale"]`.

# ⭐ Part H — Merging (with Worked Examples)
---

## Setup: Build Two Related DataFrames

````python
# DataFrame 1: student info
students = pd.DataFrame({
    "student_id": [101, 102, 103, 104],
    "name":       ["Alice", "Bob", "Carol", "Dan"],
    "major":      ["CS", "Math", "Bio", "CS"]
})

# DataFrame 2: their grades
grades = pd.DataFrame({
    "student_id": [101, 102, 103, 105],
    "course":     ["Calc", "Stats", "Bio101", "Calc"],
    "grade":      [95, 87, 91, 78]
})

print("Students:")
print(students)
print()
print("Grades:")
print(grades)
````

Notice: **Dan (104)** is in students but has no grade. **Student 105** has a grade but isn't in students. This will let us see how `how="left"` vs `how="right"` differs.

# 28. `pd.merge()` (left/right/how/on)

## `how="left"` — Keep all LEFT rows

````python
merged_left = pd.merge(
    left  = students,
    right = grades,
    how   = "left",
    on    = "student_id"
)
print(merged_left)
````

### What happened
- All 4 students appear (Alice, Bob, Carol, Dan)
- Dan (student_id 104) has NaN for course and grade because he wasn't in `grades`
- Student 105 from `grades` was DROPPED because they're not in `students`

## `how="right"` — Keep all RIGHT rows

````python
merged_right = pd.merge(
    left  = students,
    right = grades,
    how   = "right",
    on    = "student_id"
)
print(merged_right)
````

### What happened
- All 4 grades appear, including student 105
- Student 105 has NaN for name and major because they weren't in `students`
- Dan (student 104) was DROPPED because he had no grade

## Subsetting Columns Before Merging

````python
merged_subset = pd.merge(
    left  = students[["student_id", "name"]],          # only id and name
    right = grades[["student_id", "grade"]],           # only id and grade
    how   = "left",
    on    = "student_id"
)
print(merged_subset)
````

## Controlling Column Order

````python
# Want columns in order: name, student_id, major, grade
# Solution: put 'name' FIRST in the left subset

reordered = pd.merge(
    left  = students[["name", "student_id", "major"]],
    right = grades[["student_id", "grade"]],
    how   = "left",
    on    = "student_id"
)
print(reordered)
````

> ✅ **Key insight:** column order in the result follows: (1) the LEFT DataFrame's column order, then (2) the RIGHT DataFrame's column order — **excluding** the `on` column which only appears once. This is exactly what Practice Quiz 4 Q3 tests!

## Multi-Column Merge

````python
# Sometimes you need to match on TWO columns at once
orders = pd.DataFrame({
    "order_id":   [1, 1, 2, 2],
    "product_id": ["A", "B", "A", "C"],
    "qty":        [3, 1, 2, 5]
})

prices = pd.DataFrame({
    "order_id":   [1, 1, 2, 2],
    "product_id": ["A", "B", "A", "C"],
    "price":      [10, 20, 12, 8]
})

merged_multi = pd.merge(
    left  = orders,
    right = prices,
    how   = "left",
    on    = ["order_id", "product_id"]   # LIST of keys!
)
print(merged_multi)
````

# 29. Positional vs Keyword Arguments

````python
# Define a function with 4 parameters
def invest(P, r, n, t):
    V = P * (1 + r/n) ** (n*t)
    return round(V, 2)
````

## All positional (✅ Valid)

````python
# Order matters! P=1000, r=0.05, n=12, t=10
result = invest(1000, 0.05, 12, 10)
print(result)
````

## All keyword (✅ Valid)

````python
result = invest(P=1000, r=0.05, n=12, t=10)
print(result)
````

## Keywords reordered (✅ Valid)

````python
# Order doesn't matter when you use keywords!
result = invest(t=10, P=1000, n=12, r=0.05)
print(result)
````

## Mixed: positional first, then keyword (✅ Valid)

````python
# 1000 -> P, 0.05 -> r, then keywords for the rest
result = invest(1000, 0.05, n=12, t=10)
print(result)
````

## Mixed: keyword first, then positional (❌ ERROR)

````python
# UNCOMMENT to see the error:
# result = invest(P=1000, 0.05, 12, 10)
# SyntaxError: positional argument follows keyword argument
````

> ⚠️ Once you use a keyword, ALL remaining arguments must also be keyword.

## Variable names DON'T need to match parameter names

````python
def add(a, b):
    return a + b

# These all work — the parameter names a, b are LOCAL to the function
x = 3
y = 7

print(add(x, y))         # x->a, y->b
print(add(y, x))         # y->a, x->b (different order!)
print(add(b=x, a=y))     # explicit assignment via keywords
````

# 30. Indexing Into Return Values

````python
def alt_quad_formula(a, b, c):
    discrim = np.sqrt(b**2 - 4*a*c)
    x1 = (2*c) / (-b - discrim)
    x2 = (2*c) / (-b + discrim)
    return x1, x2     # returns a TUPLE
````

## Choice 1: Unpack into separate variables

````python
# The number of variables on the left MUST match the tuple length
root1, root2 = alt_quad_formula(3, -3, -6)
print("root1:", root1)
print("root2:", root2)
````

## Choice 2: Store as a tuple, index into it

````python
output = alt_quad_formula(3, -3, -6)

print("Type:", type(output))
print("Full output:", output)
print("output[0]:", output[0])
print("output[1]:", output[1])
````

## Ignoring a value with `_`

````python
# If you only care about one root
_, root2 = alt_quad_formula(3, -3, -6)
print("root2:", root2)
# The _ is just a convention meaning 'I don't care about this value'
````

## Common Mistakes

### Mistake 1: Wrong number of variables when unpacking

````python
# Function returns 2 values, but we try to unpack into 3
# UNCOMMENT to see the error:
# x, y, z = alt_quad_formula(3, -3, -6)
# ValueError: not enough values to unpack (expected 3, got 2)
````

### Mistake 2: Indexing past the end

````python
output = alt_quad_formula(3, -3, -6)
# UNCOMMENT to see the error:
# print(output[2])
# IndexError: tuple index out of range
````

### Mistake 3: Forgetting parentheses when calling

````python
# Without parentheses, you get the function ITSELF, not its result
result = alt_quad_formula
print(result)
print(type(result))

# With parentheses, you get the actual result
result = alt_quad_formula(3, -3, -6)
print(result)
print(type(result))
````

## Calling functions in different contexts

### Inside `print()`

````python
# The inner function runs first, then its return value is printed
print(alt_quad_formula(3, -3, -6))
print(invest(1000, 0.05, 12, 10))
````

### Inside `.apply()` (no parentheses!)

````python
# Pass the function NAME, no parens — Pandas calls it for you
df_test = pd.DataFrame({"x": [1, 2, 3, 4, 5]})

def double(value):
    return value * 2

df_test["doubled"] = df_test["x"].apply(double)   # NO parens on double!
print(df_test)
````

### Inside other functions (with parentheses)

````python
def square(x):
    return x ** 2

def sum_of_squares(a, b, c):
    # Call square(a), square(b), square(c) and add them
    return square(a) + square(b) + square(c)

print(sum_of_squares(1, 2, 3))   # 1 + 4 + 9 = 14
````

## Pandas Operations

| Task | Code | Returns |
|------|------|---------|
| Read CSV | `pd.read_csv("file.csv")` | DataFrame |
| Preview | `df.head(n)` | DataFrame |
| Column names | `df.columns.values` | NumPy array |
| Data types | `df.dtypes` | Series |
| Single column | `df["col"]` | Series |
| Multiple columns | `df[["a","b"]]` | DataFrame |
| Position-based | `df.iloc[r, c]` | Value/Series/DF |
| Sort | `df.sort_values(by="col")` | DataFrame |
| Filter | `df.query("col > val")` | DataFrame |
| Filter w/ var | `df.query("col > @v")` | DataFrame |
| Spaced col | `` df.query("`col name` > 5") `` | DataFrame |
| Row count | `len(df)` or `df.shape[0]` | Integer |
| Add column | `df["new"] = values` | Modifies df |
| Apply func | `series.apply(func)` | Series |
| Drop column | `df.drop(columns="c", inplace=True)` | Modifies df |
| Rename | `df.rename(columns={"old":"new"}, inplace=True)` | Modifies df |
| Save CSV | `df.to_csv("file.csv", index=False)` | Writes file |

## Cleaning, Recoding, Grouping

| Task | Code |
|------|------|
| Check dtype | `df.dtypes` |
| String numeric check | `df["col"].str.isnumeric()` |
| Replace bad values | `df["col"].replace(old_list, new_list)` |
| To numeric (strict) | `pd.to_numeric(df["col"])` |
| To numeric (auto-NaN) | `pd.to_numeric(df["col"], errors="coerce")` |
| Recode w/ apply | `df["new"] = df["old"].apply(my_func)` |
| Recode w/ cut | `pd.cut(df["col"], bins=[...], labels=[...], right=True)` |
| Robust bins | `bins = [-np.inf, 20, 34, np.inf]` |
| Group by col | `df.groupby(by="col")` |
| Group + mean | `df.groupby("col")["target"].mean()` |
| Multi-col group | `df.groupby(["col1","col2"])` |
| Multiple stats | `df["col"].agg(["mean","std","min","max"])` |
| Named agg | `df.agg(my_mean=('col','mean'))` |
| Group + agg | `df.groupby('g').agg(avg=('x','mean'))` |
| Filter→group→agg | `df.query(...).groupby(...).agg(...)` |
